```python
import pandas as pd
import numpy as np
```

 1 creating a dataframe from array 

1.1 option 1 

```python
# creating an array
data = np.array([[1,3],[2,5],[3,6]])
```

```python
# Creating a dataframe
df = pd.DataFrame(data, index=['row1', 'row2', 'row3'],
                 columns=['col1', 'col2'])
df
```

1.2 option 2 

```python
# Creating a array with list
data = [[1,3],[2,5],[3,6]]
```

```python
# Creating a dataframe 
df = pd.DataFrame(data, index=['row1', 'row2', 'row3'],
                 columns=['col1', 'col2'])
df
```

2. Creating data frame using dictionary 

```python
# lists
dept = ['civil', 'mech', 'eee', 'ece', 'tech']
students = [30, 20, 25, 40, 60]
```

```python
dict = {'Dept':dept , 'Students':students}
```

```python
df = pd.DataFrame(dict)
df
```
