```python
import pandas as pd
df = pd.read_csv('organizations.csv')
df.head()
```

1 Create Index

```python
import numpy as np 
import random
```

```python
# creating a non repetive value for the index
new_index = np.arange(0,1000)
```

```python
# shuffling the indexes
random.shuffle(new_index)
```

```python
# creating a new column with new index
df.insert(1, 'New_Index', new_index)
```

```python
df
```

Set Index

```python
# Set new_index column as index
df.set_index('New_index', inplace=True)
df
```

3. Sort Index 

```python
# sort data frame by new index
df.sort_index()
```

```python
# sort descending by index
df.sort_index(ascending=False)
```
