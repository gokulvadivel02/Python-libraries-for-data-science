```python
import pandas as pd
df = pd.read_csv('organizations.csv')
df.head()
```

1. Sort a dataframe

```python
# Sort by column
df.sort_values(by='Founded')
```

```python
# sort descending by  one column
df.sort_values(by='Founded', ascending=False)
```

```python
# sort descending by multiple column
df.sort_values(['Founded', 'Number of employees'], ascending=False)
```

```python
# sort descending by multipe=le columns and update original dataframe
df.sort_values(['Founded', 'Number of employees'], ascending=False,
               inplace=True)
df
```

```python
df.sort_values(['Country'], ascending=False)
```
