```python
import pandas as pd
import numpy as np
df = pd.read_csv('people.csv')
```

1. assign() :- using assign function for adding multiple values

```python
# creating an new arrays with random numbers
score1 = np.random.randint(1,100, size=1000)
score2 = np.random.randint(1,100, size=1000)
```

```python
# Creating a series using the random numbers 
s1 = pd.Series(score1, index=np.arange(0,1000))
s2 = pd.Series(score2, index=np.arange(0,1000))
```

```python
# using assign() to add multiple columns
# assing method only create a copy overide using assign it to original dataframe
df = df.assign(Score1 = s1, Score2= s2)
df
```

2. insert() :- insert method used to insert a column at specfic position or index

```python
# using insert() to add a column to the specfic index
# syntax :- insert(indes value, name of the column, value)
df.insert(2, 'random', s2)
df
```
