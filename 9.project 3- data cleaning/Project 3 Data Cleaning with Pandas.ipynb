# Identifying Missing Data

```python
df.isnull().sum().sort_values(ascending=False)
```

```python
for column in df.columns:
    percentage=df[column].isnull().mean()
    print(f"{column}:{round(percentage*100, 2)}%")
```

# Dealing with missing data

```python
# Remove a column or row with .drop, .dropna or .isnull

# drop column
df.drop('director', axis=1).head()
```

```python
director_null = df[df['director'].isnull()].index
df.drop(director_null, axis=0).head()
```

```python
df.dropna(subset=['director']).head()
```

```python
df[df['rating'].isnull()]
mode = ''.join(df['rating'].mode())
df['rating'].fillna(mode, inplace=True)
```

```python
df[df['rating'].isnull()]
```

```python
df[df['duration'].isnull()]
```

```python
df['duration'].fillna(0,inplace=True)

# df.fillna(method='ffill')
# df.fillna(method='bfill')
```

```python
df[df['duration'].isnull()]
```
