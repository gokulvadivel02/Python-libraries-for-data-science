## all links

- https://www.football-data.co.uk/mmz4281/2526/E1.csv
- https://www.football-data.co.uk/mmz4281/2526/E2.csv
- https://www.football-data.co.uk/mmz4281/2526/E3.csv
- https://www.football-data.co.uk/mmz4281/2526/EC.csv

```python
# creating the root variable (common for all link)
root = "https://www.football-data.co.uk/mmz4281/"
```

```python
# Storing only season data
# creating a list with dynamic values (this is diff in all link)
leagues = ['E0','E1','E2','E3','EC']
frames=[] 

# looping the league and appending in frames list
for league in leagues:
    df = pd.read_csv(root +'2526'+'/'+league+'.csv')
    frames.append(df)
```

```python
# length of the frames
len(frames)
```

# Multiple Season Data

```python
for season in range(20,26):
    print(str(season)+str(season+1))
```

```python
# Multiple seasons
leagues = ['E0','E1','E2']
frames=[] 

for league in leagues:
    for season in range(20,26):
        df = pd.read_csv(root +str(season)+str(season+1)+'/'+league+'.csv')
        df.insert(1,'Season', season)
        frames.append(df)
```

```python
# creating a dictionary 
dict_cont={
    'Spanish la ': 'SP1', 'Spanish SD' : 'SP2',
    'German' : 'D1', 'English prem':'E0', 'English league':'E2', 
    'English':'E3'
}
```

```python
for league in dict_cont:
    print(dict_cont[league])
```

```python
# storing specfic key of data 
# leagues = ['E0','E1','E2']
total_dict={}

for league in dict_cont:
    frames=[] 
    for season in range(20,26):
        df = pd.read_csv(root +str(season)+str(season+1)+'/'+ dict_cont[league] +'.csv')
        df.insert(1,'Season', season)
        frames.append(df)
    df_concat = pd.concat(frames)
    total_dict[league] = df_concat 
    
        
```
