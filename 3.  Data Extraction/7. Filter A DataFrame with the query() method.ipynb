```python
import pandas as pd
```

```python
# reads csv file
df = pd.read_csv('players_20.csv')
# set index
df.set_index('short_name', inplace=True)
# select column
df = df[['long_name', 'age', 'dob', 'height_cm', 'weight_kg', 'nationality', 'club']]
```

```python
df.head()
```

# query()

```python
df.query('age>34')

df[df['age']>34].head()
```

```python
df.query("age>34 and nationality == 'Portugal' ")

df[(df['age']>34) & (df['nationality']=='Portugal')]
```

```python
df.query('not(age>34)')

df[~(df['age']>34)]
```

```python
df.query('height_cm/100 == 1.8')
df[df['height_cm']/100 == 1.8]
```

```python
# select players that were born after 1990

# check out data types
df.dtypes

# convert "dob" column to datetime type
df['dob'] = df['dob'].astype('datetime64[ns]')
df['dob'].dtypes

# query
df.query('dob.dt.year > 1990')

df[df['dob'].dt.year >1990]
```
