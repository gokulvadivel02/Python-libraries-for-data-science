```python
import pandas as pd
```

```python
# reads csv file
df = pd.read_csv('players_20.csv')
# set index
df.set_index('short_name', inplace=True)
# select column
df = df[['long_name', 'age', 'dob', 'height_cm', 'weight_kg', 'nationality', 'club']]
```

```python
df.head(3)
```

# sample()

```python
# give some sample rows 0.2=> 20% of rows of original df
# random_state=50 gives same sample if we run how many times also (99, 50 oru wish)
df.sample(frac=0.2, random_state=50)
```

```python
# frac=2 => 200% of sample double the original df
# replace=True => for 200% of data the repetation of row required so replace=True allows to make repetaion
# random_state=99 => give same sample no matter how many times we run
df.sample(frac=2, replace=True, random_state=99)
```
