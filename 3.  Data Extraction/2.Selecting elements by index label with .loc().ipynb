```python
import pandas as pd
```

```python
# reads csv file
df = pd.read_csv('players_20.csv')
# set index
df.set_index('short_name', inplace=True)
# select column
df = df[['long_name', 'age', 'dob', 'height_cm', 'weight_kg', 'nationality', 'club']]
```

```python
df.head(3)
```

# Selecting with a single value

syntax :- loc[row_label, column_label]

```python
# geting all data of individual row 
df.loc['L. Messi']
```

```python
# getting specfic data from a row 
df.loc['L. Messi','club']
# get the data of messi`s club name 
```

```python
df.loc['Neymar Jr','dob' ]
# getting nymer`s date ob birth
```

```python
# get all rows of a specfic column 
df.loc[:,'nationality'].head(3)
```

```python
# get all data of a row or data of nymar 
df.loc['Neymar Jr']
```

# Selecting with a list of values

```python
df.loc[['Cristiano Ronaldo', 'Neymar Jr']]
```

```python
df.loc[['L. Messi', 'Neymar Jr'],'height_cm']
```

```python
df.loc[['L. Messi', 'Cristiano Ronaldo', 'Neymar Jr'],['age', 'height_cm','weight_kg']]
```

# Selecting a range of data with a slice

start:stop:step (Note that contrary to usual python slices, both the start and the stop are included)

```python
players = ['L. Messi', 'Cristiano Ronaldo']
# slicing the column 
df.loc[players, 'age':'nationality' ]
```

```python
# slice index labels
columns = ['age', 'dob', 'height_cm', 'weight_kg']
# slicing first 10 players with required columns
df.loc['L. Messi':'M. Salah', columns]
```

# Selecting with conditions

```python
# selecting the players age 40 above 
columns = ['age', 'dob', 'height_cm', 'weight_kg'] 
df.loc[df['age']>40, columns]
```

```python
df.loc[(df['height_cm']>180) & (df['weight_kg']>80), :]
```
