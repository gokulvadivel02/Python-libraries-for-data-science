```python
import pandas as pd
```

```python
df_laptop = pd.read_csv('laptop_price.csv', encoding='unicode_escape')
```

```python
df_laptop.head(2)
```

# filtering a data frame 

```python
df_laptop.value_counts('Company')
```

```python
df_laptop['Price_euros']<2000
```

```python
df_laptop['Company']=='Lenovo'
```

# using and '&' operator for filtering

```python
df_laptop[(df_laptop['Price_euros']>2000) & (df_laptop['Company']=='Lenovo')]
## this will display on;y the lenovo laptop with price above 2000
```

# using or '|' operator for filtering

```python
df_laptop[(df_laptop['Price_euros']>2000) | (df_laptop['Company']=='Lenovo')]
# this will display all the laptop price is above 2000 
# and all lenovo laptop no matter what price 
```
