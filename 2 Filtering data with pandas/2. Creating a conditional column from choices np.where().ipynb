# Creating a Conditional column from 2 chocies np.where()

syntax :- np.where(condition, value_if_true, value_if_false)

```python
import numpy as np 
```

```python
# Creating a array based on price .
np.where(df_laptop['Price_euros']>2000, 'Expensive', 'Inexpensive')
```

```python
# adding a new column to the dataframe
df_laptop['Price_desc'] = np.where(df_laptop['Price_euros']>2000,'Expensive', 'Inexpensive')
```

```python
df_laptop.head(2)
```

```python
# counting the no of values in Price_desc
df_laptop.value_counts('Price_desc')
```
