```python

import pandas as pd
```

```python
df_laptop = pd.read_csv('laptop_price.csv', encoding='unicode_escape')
```

```python

df_laptop.head(2)
```

# drop_duplicates()

```python
# default keep = first that means keep all first occrence and delete remaining all
df_laptop.drop_duplicates(['Company'])
df_laptop.drop_duplicates(['Company'])[['Company', 'Price_euros']]
```

```python
# sorting the df in ascending order default ascending = True
df_laptop = df_laptop.sort_values(['Company', 'Price_euros'])
```

```python
# (keep = first ) as default 
df_laptop.drop_duplicates(['Company'], keep='first')[['Company', 'Price_euros']]
```
