```python
import pandas as pd 
```

```python
df_laptop = pd.read_csv('Laptop_price.csv', encoding='unicode_escape')
```

```python
df_laptop.head()
```

```python
# finding which row consist lenovo in the company column
lenovo = df_laptop['Company'] == 'Lenovo'
lenovo
```

```python
# filtering the df only company that contain lenovo
df_laptop[lenovo].head()
```

```python
df_laptop[lenovo].value_counts('Company')
```

```python
# companies withot hp containing rows
not_hp = df_laptop['Company']!='hp'
```

```python
df_laptop[not_hp]
```

```python
# count of laptop other the hp
df = df_laptop[not_hp].value_counts('Company')
df
```
