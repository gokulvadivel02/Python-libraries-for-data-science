 Task: Find the biggest and smallest screen size in laptops of each company using the sort_values() and duplicated() methods (keep "first", "last" and False)

```python
# sort dataframe asending by company and inches
df.sort_values(['Inches','Company'])
```

```python
# smallest: keep='first'
sort = df.sort_values(['Company', 'Inches'])
sort.drop_duplicates(['Company'])[['Company', 'Inches']]
```

```python
sort.drop_duplicates(['Company'], keep='last')[['Company', 'Inches']]
```
