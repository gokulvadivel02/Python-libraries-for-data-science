```python
import pandas as pd
```

```python
df_laptop = pd.read_csv('laptop_price.csv', encoding='unicode_escape')
```

```python
df_laptop.head(2)
```

# unique()


Return unique values of Series object as array. (Uniques are returned in order of appearance.)

```python
df_laptop = df_laptop.sort_values(['Company', 'Inches'])
```

```python
df_laptop['Company'].unique()
```

```python
df_laptop['Inches'].unique()
```

# nunique()

```python
len(df_laptop['Company'].unique())
```

Return number of unique elements in the object (Excludes NA values by default)

```python
df_laptop['Company'].nunique()
```

```python
df_laptop['Inches'].nunique()
```
