# Example 2

Finding the cheapest and expensive laptop in the data frame  

```python
# sort based company and price low to high
df_laptop = df_laptop.sort_values(['Company', 'Price_euros'])
df_laptop
```

```python
# duplicate , default first-> false
duplicate_first = df_laptop.duplicated(['Company'])
```

```python
# all duplicated values & filter all value except first value 
# except first occrence all is true 
df_laptop[duplicate_first]

# using not operator '~' changing true->false, false->true
df_laptop[~duplicate_first]
# only display two column 'Company'&'Price_euros'
# display chepest price
df_laptop[['Company', 'Price_euros']]
df_laptop[~duplicate_first][['Company', 'Price_euros']]
```

```python
# set all true except last occrence last is false
duplicate_last = df_laptop.duplicated('Company', keep='last')
duplicate_last
```

```python
# set all true except the value which occrus only once or unique value
duplicate_false = df_laptop.duplicated('Company', keep=False)
duplicate_false
```

```python
# seting all last occrence or expensive as true & display only 2 column 
df_laptop[['Company', 'Price_euros']]
df_laptop[~duplicate_last][['Company', 'Price_euros']]
```
