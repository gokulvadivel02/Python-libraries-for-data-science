```python
import pandas as pd
```

```python
df = pd.read_csv('population_total.csv')
```

```python
df
```

# Making a Pivot Table

.pivot(): Returns reshaped DataFrame organized by given index / column values ("pivot without aggregation")

```python
df.dropna(inplace=True)
```

```python
df_pivot=df.pivot(index='year', columns='country', values='population')
```

```python
df_pivot=df_pivot[['India', 'United States', 'China', 
                          'Indonesia', 'Brazil']]
```

```python
df_pivot.head()
```

```python
df_pivot.plot(kind='line', 
             xlabel='year', ylabel='population',
             title='population (1955-2020)',
             figsize=(8,4))
```

# BarChart

## Single Barplot

```python
# selecting only one year 
df_2020 = df_pivot[df_pivot.index.isin([2020])]
```

```python
df_2020
```

```python
# transpose dataframe from rows to columns or viceversa
df_2020_t = df_2020.T
```

```python
df_2020_t.plot(kind='bar',color='blue',
            xlabel='year', ylabel='population',
            title='population 2020')
```

```python
df_year=df_pivot[df_pivot.index.isin([1980, 1990,2000,2010,2020])]
```

```python
df_year.plot(kind='bar',
            xlabel='year', ylabel='population',
             figsize=(8,4)
            )
```

# Piechart
